```python
import sympy as sp
import numpy as np
from IPython.display import display
```

```python
t = sp.symbols('t') #define t as symbol
x = sp.Function('x') #define x as func
y = sp.Function('y') #define y as func
z = sp.Function('z') #define z as func
a,b,c = 0.2, 0.2, 5.7 #define variables abc for Rossler attractor:unsure if abc are supposed to be linspace lists but these are values found online, can change if need be
alpha = np.linspace(0, 5, 101) #define variables for Lotka-Volterra equations
beta = np.linspace(0, 5, 101)
delta = np.linspace(0, 5, 101)
gamma = np.linspace(0, 5, 101)
```

```python
def fxfy2ndord(t, u): #define Lotka-Volterra equations
    x, y = u
    dxdt2 = alpha*x - beta*x*y #dx/dt = ax-bxy
    dydt2 = delta*x*y -gamma*y #dy/dt = cxy-dy
    return[dxdt2, dydt2]

def fxfy3rdord(t, v): #define Rossler attractor
    x, y, z = v
    dxdt3 = -y-z
    dydt3 = x + a*y
    dzdt3 = b + z*(x-c)
```
